```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
import seaborn as sns 
```

```python
import os
path = os.path.expanduser("~/Downloads/ride_hailing_dataset.csv")
df = pd.read_csv(path)
df.head()
```

```python
print("\nQ1: Average fare by service type")
avg_fare = df.groupby("service_type")["fare"].mean().round(2).sort_values(ascending=False)
print(avg_fare)
```

```python
print("\nQ2: Trips per city")
print(df["city"].value_counts())
print("Answer:", df["city"].value_counts().idxmax())
```

```python
round(df["distance_km"].mean(), 2)
```

```python
print("nQ4: Average trip rating by gender")
rating_by_gender = df.groupby("gender")["trip_rating"].mean().round(2)
print(rating_by_gender)

plt.figure(figsize=(6, 5))
sns.barplot(x=rating_by_gender.index, y=rating_by_gender.values)
plt.title("Average Trip Rating by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Trip Rating")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.barplot(x=avg_fare.index, y=avg_fare.values)
plt.title("Average Fare by Service Type")
plt.xlabel("Service Type")
plt.ylabel("Average Fare (PKR)")
plt.tight_layout()
```
